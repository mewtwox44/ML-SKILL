```python
# ============================================================
# PROJECT 5
# FULL LOGISTIC REGRESSION PIPELINE
# Dataset: Titanic - Kaggle
# Model: Logistic Regression
# ============================================================


# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    roc_auc_score,
    roc_curve
)


# ============================================================
# 2. UPLOAD DATASET
# ============================================================

print("Please upload your Kaggle Titanic train.csv file.")

uploaded = files.upload()

filename = list(uploaded.keys())[0]

df = pd.read_csv(filename)

print("\nDataset uploaded successfully!")
print("File name:", filename)
print("Dataset shape:", df.shape)


# ============================================================
# 3. DISPLAY FIRST 5 ROWS
# ============================================================

print("\nFirst 5 rows:")
display(df.head())


# ============================================================
# 4. DATASET INFORMATION
# ============================================================

print("\nDataset Information:")
df.info()


# ============================================================
# 5. STATISTICAL SUMMARY
# ============================================================

print("\nStatistical Summary:")
display(df.describe())


# ============================================================
# 6. CHECK DUPLICATES
# ============================================================

print("\nNumber of duplicate rows:")
print(df.duplicated().sum())


# ============================================================
# 7. CHECK MISSING VALUES
# ============================================================

print("\nMissing Values:")

missing = df.isnull().sum()

missing = missing[missing > 0].sort_values(ascending=False)

display(missing)


# ============================================================
# 8. TARGET VARIABLE
# ============================================================

target = "Survived"

print("\nTarget variable:", target)

print("\nTarget distribution:")
print(df[target].value_counts())


# ============================================================
# 9. TARGET DISTRIBUTION GRAPH
# ============================================================

plt.figure(figsize=(6, 5))

sns.countplot(
    x=df[target]
)

plt.xlabel("Survived")
plt.ylabel("Number of Passengers")

plt.title("Survival Distribution")

plt.show()


# ============================================================
# 10. SURVIVAL BY GENDER
# ============================================================

plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="Sex",
    hue="Survived"
)

plt.xlabel("Gender")
plt.ylabel("Number of Passengers")

plt.title("Survival by Gender")

plt.show()


# ============================================================
# 11. SURVIVAL BY PASSENGER CLASS
# ============================================================

plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="Pclass",
    hue="Survived"
)

plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")

plt.title("Survival by Passenger Class")

plt.show()


# ============================================================
# 12. SELECT FEATURES
# ============================================================

features = [
    "Pclass",
    "Sex",
    "Age",
    "SibSp",
    "Parch",
    "Fare",
    "Embarked"
]

X = df[features]

y = df[target]


print("\nSelected features:")
print(features)


# ============================================================
# 13. HANDLE MISSING NUMERICAL VALUES
# ============================================================

print("\nMissing values before preprocessing:")

display(X.isnull().sum())


# Fill Age using median

X["Age"] = X["Age"].fillna(
    X["Age"].median()
)


# Fill Fare using median

X["Fare"] = X["Fare"].fillna(
    X["Fare"].median()
)


# Fill Embarked using mode

X["Embarked"] = X["Embarked"].fillna(
    X["Embarked"].mode()[0]
)


print("\nMissing values after handling:")

display(X.isnull().sum())


# ============================================================
# 14. ENCODE CATEGORICAL FEATURES
# ============================================================

X = pd.get_dummies(
    X,
    columns=["Sex", "Embarked"],
    drop_first=True
)


print("\nFeatures after encoding:")

display(X.head())


# ============================================================
# 15. TRAIN-TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)


print("\nTraining data shape:", X_train.shape)
print("Testing data shape:", X_test.shape)


# ============================================================
# 16. FEATURE SCALING
# ============================================================

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(
    X_train
)

X_test_scaled = scaler.transform(
    X_test
)


# ============================================================
# 17. CREATE LOGISTIC REGRESSION MODEL
# ============================================================

logistic_model = LogisticRegression(
    max_iter=1000
)


# ============================================================
# 18. TRAIN MODEL
# ============================================================

logistic_model.fit(
    X_train_scaled,
    y_train
)


# ============================================================
# 19. MAKE PREDICTIONS
# ============================================================

y_pred = logistic_model.predict(
    X_test_scaled
)


# Probability of survival

y_probability = logistic_model.predict_proba(
    X_test_scaled
)[:, 1]


# ============================================================
# 20. ACCURACY
# ============================================================

accuracy = accuracy_score(
    y_test,
    y_pred
)

print("\n========================================")
print("MODEL ACCURACY")
print("========================================")

print("Accuracy:", accuracy)


# ============================================================
# 21. PRECISION
# ============================================================

precision = precision_score(
    y_test,
    y_pred
)

print("\nPrecision:", precision)


# ============================================================
# 22. RECALL
# ============================================================

recall = recall_score(
    y_test,
    y_pred
)

print("\nRecall:", recall)


# ============================================================
# 23. F1 SCORE
# ============================================================

f1 = f1_score(
    y_test,
    y_pred
)

print("\nF1 Score:", f1)


# ============================================================
# 24. ROC-AUC
# ============================================================

roc_auc = roc_auc_score(
    y_test,
    y_probability
)

print("\nROC-AUC:", roc_auc)


# ============================================================
# 25. CLASSIFICATION REPORT
# ============================================================

print("\n========================================")
print("CLASSIFICATION REPORT")
print("========================================")

print(
    classification_report(
        y_test,
        y_pred
    )
)


# ============================================================
# 26. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    y_test,
    y_pred
)

print("\nConfusion Matrix:")
print(cm)


# ============================================================
# 27. CONFUSION MATRIX HEATMAP
# ============================================================

plt.figure(figsize=(6, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues"
)

plt.xlabel("Predicted Label")
plt.ylabel("Actual Label")

plt.title("Confusion Matrix")

plt.show()


# ============================================================
# 28. ROC CURVE
# ============================================================

fpr, tpr, thresholds = roc_curve(
    y_test,
    y_probability
)


plt.figure(figsize=(8, 6))

plt.plot(
    fpr,
    tpr,
    label=f"Logistic Regression (AUC = {roc_auc:.3f})"
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--"
)

plt.xlabel("False Positive Rate")

plt.ylabel("True Positive Rate")

plt.title("ROC Curve")

plt.legend()

plt.show()


# ============================================================
# 29. MODEL COEFFICIENTS
# ============================================================

coefficients = pd.DataFrame({

    "Feature": X.columns,

    "Coefficient": logistic_model.coef_[0]

})


print("\n========================================")
print("LOGISTIC REGRESSION COEFFICIENTS")
print("========================================")

display(coefficients)


# ============================================================
# 30. COEFFICIENT VISUALIZATION
# ============================================================

plt.figure(figsize=(10, 6))

plt.barh(
    coefficients["Feature"],
    coefficients["Coefficient"]
)

plt.xlabel("Coefficient Value")

plt.ylabel("Feature")

plt.title("Logistic Regression Coefficients")

plt.show()


# ============================================================
# 31. ACTUAL VS PREDICTED
# ============================================================

comparison = pd.DataFrame({

    "Actual": y_test.values,

    "Predicted": y_pred,

    "Survival Probability": y_probability

})


print("\n========================================")
print("ACTUAL VS PREDICTED")
print("========================================")

display(comparison.head(20))


# ============================================================
# 32. MODEL PERFORMANCE TABLE
# ============================================================

performance = pd.DataFrame({

    "Metric": [
        "Accuracy",
        "Precision",
        "Recall",
        "F1 Score",
        "ROC-AUC"
    ],

    "Score": [
        accuracy,
        precision,
        recall,
        f1,
        roc_auc
    ]

})


print("\n========================================")
print("MODEL PERFORMANCE")
print("========================================")

display(performance)


# ============================================================
# 33. FINAL CONCLUSION
# ============================================================

print("\n========================================")
print("FINAL CONCLUSION")
print("========================================")

print("""
A complete Logistic Regression pipeline was implemented
using the Titanic dataset.

The dataset was first explored and missing values were
handled using median and mode imputation.

Categorical variables were converted into numerical
variables using one-hot encoding.

Numerical features were standardized using StandardScaler.

Logistic Regression was then trained to predict whether
a passenger survived.

The model was evaluated using Accuracy, Precision,
Recall, F1 Score, Confusion Matrix and ROC-AUC.

These evaluation metrics help determine the classification
performance of the Logistic Regression model.
""")
```
