```python
# ============================================================
# PROJECT 4
# LINEAR REGRESSION AND REGULARISATION
# Dataset: House Prices - Kaggle
# Models:
# 1. Linear Regression
# 2. Ridge Regression
# 3. Lasso Regression
# ============================================================


# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score


# ============================================================
# 2. UPLOAD DATASET
# ============================================================

print("Please upload your Kaggle House Prices train.csv file.")

uploaded = files.upload()

filename = list(uploaded.keys())[0]

df = pd.read_csv(filename)

print("\nDataset uploaded successfully!")
print("File name:", filename)
print("Dataset shape:", df.shape)


# ============================================================
# 3. DISPLAY FIRST 5 ROWS
# ============================================================

print("\nFirst 5 rows:")
display(df.head())


# ============================================================
# 4. DATASET INFORMATION
# ============================================================

print("\nDataset Information:")
df.info()


# ============================================================
# 5. STATISTICAL SUMMARY
# ============================================================

print("\nStatistical Summary:")
display(df.describe())


# ============================================================
# 6. CHECK DUPLICATES
# ============================================================

print("\nNumber of duplicate rows:")
print(df.duplicated().sum())


# ============================================================
# 7. CHECK MISSING VALUES
# ============================================================

print("\nMissing Values:")

missing = df.isnull().sum()

missing = missing[missing > 0].sort_values(ascending=False)

display(missing)


# ============================================================
# 8. TARGET VARIABLE
# ============================================================

target = 'SalePrice'

print("\nTarget variable:", target)

print("\nSalePrice statistics:")
display(df[target].describe())


# ============================================================
# 9. DISTRIBUTION OF SALE PRICE
# ============================================================

plt.figure(figsize=(8, 5))

plt.hist(df[target], bins=30)

plt.xlabel("Sale Price")
plt.ylabel("Frequency")
plt.title("Distribution of SalePrice")

plt.show()


# ============================================================
# 10. SELECT FEATURES
# ============================================================

features = [
    'OverallQual',
    'GrLivArea',
    'GarageCars',
    'GarageArea',
    'TotalBsmtSF',
    '1stFlrSF',
    'FullBath',
    'YearBuilt',
    'TotRmsAbvGrd'
]

X = df[features]

y = df[target]


print("\nSelected Features:")
print(features)


# ============================================================
# 11. CHECK MISSING VALUES IN SELECTED FEATURES
# ============================================================

print("\nMissing values in selected features:")

display(X.isnull().sum())


# ============================================================
# 12. HANDLE MISSING VALUES
# ============================================================

X = X.fillna(X.median())

print("\nMissing values after handling:")

display(X.isnull().sum())


# ============================================================
# 13. CORRELATION HEATMAP
# ============================================================

correlation_data = df[features + [target]].corr()

plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation_data,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.show()


# ============================================================
# 14. GRLIVAREA VS SALEPRICE
# ============================================================

plt.figure(figsize=(8, 5))

plt.scatter(
    df['GrLivArea'],
    df['SalePrice']
)

plt.xlabel("Above Ground Living Area")
plt.ylabel("Sale Price")

plt.title("GrLivArea vs SalePrice")

plt.show()


# ============================================================
# 15. TRAIN-TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\nTraining data shape:", X_train.shape)
print("Testing data shape:", X_test.shape)


# ============================================================
# 16. FEATURE SCALING
# ============================================================

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)


# ============================================================
# 17. LINEAR REGRESSION
# ============================================================

linear_model = LinearRegression()

linear_model.fit(
    X_train_scaled,
    y_train
)

y_pred_linear = linear_model.predict(
    X_test_scaled
)


# ============================================================
# 18. LINEAR REGRESSION EVALUATION
# ============================================================

mae_linear = mean_absolute_error(
    y_test,
    y_pred_linear
)

mse_linear = mean_squared_error(
    y_test,
    y_pred_linear
)

rmse_linear = np.sqrt(mse_linear)

r2_linear = r2_score(
    y_test,
    y_pred_linear
)


print("\n========================================")
print("LINEAR REGRESSION RESULTS")
print("========================================")

print("MAE  :", mae_linear)
print("MSE  :", mse_linear)
print("RMSE :", rmse_linear)
print("R2   :", r2_linear)


# ============================================================
# 19. RIDGE REGRESSION
# ============================================================

ridge_model = Ridge(alpha=10)

ridge_model.fit(
    X_train_scaled,
    y_train
)

y_pred_ridge = ridge_model.predict(
    X_test_scaled
)


# ============================================================
# 20. RIDGE EVALUATION
# ============================================================

mae_ridge = mean_absolute_error(
    y_test,
    y_pred_ridge
)

mse_ridge = mean_squared_error(
    y_test,
    y_pred_ridge
)

rmse_ridge = np.sqrt(mse_ridge)

r2_ridge = r2_score(
    y_test,
    y_pred_ridge
)


print("\n========================================")
print("RIDGE REGRESSION RESULTS")
print("========================================")

print("MAE  :", mae_ridge)
print("MSE  :", mse_ridge)
print("RMSE :", rmse_ridge)
print("R2   :", r2_ridge)


# ============================================================
# 21. LASSO REGRESSION
# ============================================================

lasso_model = Lasso(
    alpha=100,
    max_iter=10000
)

lasso_model.fit(
    X_train_scaled,
    y_train
)

y_pred_lasso = lasso_model.predict(
    X_test_scaled
)


# ============================================================
# 22. LASSO EVALUATION
# ============================================================

mae_lasso = mean_absolute_error(
    y_test,
    y_pred_lasso
)

mse_lasso = mean_squared_error(
    y_test,
    y_pred_lasso
)

rmse_lasso = np.sqrt(mse_lasso)

r2_lasso = r2_score(
    y_test,
    y_pred_lasso
)


print("\n========================================")
print("LASSO REGRESSION RESULTS")
print("========================================")

print("MAE  :", mae_lasso)
print("MSE  :", mse_lasso)
print("RMSE :", rmse_lasso)
print("R2   :", r2_lasso)


# ============================================================
# 23. MODEL COMPARISON TABLE
# ============================================================

results = pd.DataFrame({

    "Model": [
        "Linear Regression",
        "Ridge Regression",
        "Lasso Regression"
    ],

    "MAE": [
        mae_linear,
        mae_ridge,
        mae_lasso
    ],

    "MSE": [
        mse_linear,
        mse_ridge,
        mse_lasso
    ],

    "RMSE": [
        rmse_linear,
        rmse_ridge,
        rmse_lasso
    ],

    "R2": [
        r2_linear,
        r2_ridge,
        r2_lasso
    ]
})


print("\n========================================")
print("MODEL COMPARISON")
print("========================================")

display(results)


# ============================================================
# 24. R2 SCORE COMPARISON
# ============================================================

plt.figure(figsize=(8, 5))

plt.bar(
    results["Model"],
    results["R2"]
)

plt.xlabel("Model")
plt.ylabel("R2 Score")

plt.title("R2 Score Comparison")

plt.xticks(rotation=20)

plt.show()


# ============================================================
# 25. RMSE COMPARISON
# ============================================================

plt.figure(figsize=(8, 5))

plt.bar(
    results["Model"],
    results["RMSE"]
)

plt.xlabel("Model")
plt.ylabel("RMSE")

plt.title("RMSE Comparison")

plt.xticks(rotation=20)

plt.show()


# ============================================================
# 26. ACTUAL VS PREDICTED
# ============================================================

plt.figure(figsize=(8, 5))

plt.scatter(
    y_test,
    y_pred_linear
)

plt.xlabel("Actual Sale Price")
plt.ylabel("Predicted Sale Price")

plt.title("Actual vs Predicted Sale Price")

plt.show()


# ============================================================
# 27. MODEL COEFFICIENTS
# ============================================================

coefficients = pd.DataFrame({

    "Feature": features,

    "Linear Regression": linear_model.coef_,

    "Ridge Regression": ridge_model.coef_,

    "Lasso Regression": lasso_model.coef_

})


print("\n========================================")
print("MODEL COEFFICIENTS")
print("========================================")

display(coefficients)


# ============================================================
# 28. COEFFICIENT COMPARISON
# ============================================================

coefficients.set_index(
    "Feature"
).plot(
    kind="bar",
    figsize=(12, 6)
)

plt.title("Model Coefficient Comparison")

plt.xlabel("Features")

plt.ylabel("Coefficient Value")

plt.xticks(rotation=45)

plt.show()


# ============================================================
# 29. RIDGE REGULARISATION - DIFFERENT ALPHA VALUES
# ============================================================

ridge_alphas = [
    0.01,
    0.1,
    1,
    10,
    100,
    1000
]

ridge_scores = []


for alpha in ridge_alphas:

    model = Ridge(alpha=alpha)

    model.fit(
        X_train_scaled,
        y_train
    )

    predictions = model.predict(
        X_test_scaled
    )

    score = r2_score(
        y_test,
        predictions
    )

    ridge_scores.append(score)


ridge_results = pd.DataFrame({
    "Alpha": ridge_alphas,
    "R2 Score": ridge_scores
})


print("\n========================================")
print("RIDGE ALPHA RESULTS")
print("========================================")

display(ridge_results)


# ============================================================
# 30. RIDGE ALPHA GRAPH
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    ridge_alphas,
    ridge_scores,
    marker="o"
)

plt.xscale("log")

plt.xlabel("Alpha")

plt.ylabel("R2 Score")

plt.title("Ridge Regression - Alpha vs R2")

plt.show()


# ============================================================
# 31. LASSO REGULARISATION - DIFFERENT ALPHA VALUES
# ============================================================

lasso_alphas = [
    0.01,
    0.1,
    1,
    10,
    100,
    1000
]

lasso_scores = []


for alpha in lasso_alphas:

    model = Lasso(
        alpha=alpha,
        max_iter=10000
    )

    model.fit(
        X_train_scaled,
        y_train
    )

    predictions = model.predict(
        X_test_scaled
    )

    score = r2_score(
        y_test,
        predictions
    )

    lasso_scores.append(score)


lasso_results = pd.DataFrame({
    "Alpha": lasso_alphas,
    "R2 Score": lasso_scores
})


print("\n========================================")
print("LASSO ALPHA RESULTS")
print("========================================")

display(lasso_results)


# ============================================================
# 32. LASSO ALPHA GRAPH
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    lasso_alphas,
    lasso_scores,
    marker="o"
)

plt.xscale("log")

plt.xlabel("Alpha")

plt.ylabel("R2 Score")

plt.title("Lasso Regression - Alpha vs R2")

plt.show()


# ============================================================
# 33. FIND BEST MODEL
# ============================================================

best_model = results.loc[
    results["R2"].idxmax()
]


print("\n========================================")
print("BEST MODEL")
print("========================================")

print("Model :", best_model["Model"])
print("R2    :", best_model["R2"])
print("RMSE  :", best_model["RMSE"])
print("MAE   :", best_model["MAE"])


# ============================================================
# 34. FINAL CONCLUSION
# ============================================================

print("\n========================================")
print("FINAL CONCLUSION")
print("========================================")

print("""
Linear Regression was used as the baseline model
for predicting house prices.

Ridge Regression uses L2 regularisation to reduce
the magnitude of model coefficients and help control
overfitting.

Lasso Regression uses L1 regularisation and can reduce
some coefficients to zero, which can perform feature
selection.

The models were evaluated using MAE, MSE, RMSE and R2.

The model with the highest R2 and lower error values
provides better predictive performance on the test data.
""")
```
